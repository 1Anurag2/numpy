```python
# copy    in copy if we change in first variable then no change in copy variable 
```

```python
import numpy as np
var = np.array([4,5,7,8,9])
c = var.copy()        # create copy of var in c variable 

var[2] = 55

print(var)
print()
print(c)
```

```python
# view    in view if we change in first variable then same data is change in copy variable 
```

```python
x = np.array([4,6,2,5,2])
y = x.view()
x[2] = 55
print(x)
print()
print(y)
```
