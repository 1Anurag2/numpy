```python
import numpy as np
var = np.array([3,4,5,6,7,8,9,1])
for i in var:
    print(i)
```

```python
x = np.array([[4,5,6,7,2],[4,6,7,8,0]])
for i in x:
    print(i)
print()
for i in x:
    for j in i:
        print(j)
```

```python
# for three dimession array
```

```python
var = np.array([[[3,4,5,6],[4,3,2,1],[9,8,7,0]]])
for i in var:
    for j in i:
        for k in j:
            print(k)

print()
for i in np.nditer(var):          # nditer(variable_name)  function is used to directly iterate the array
    print(i)
print()
for i in np.nditer(var,flags = ["buffered"],op_dtypes=["S"]):          # nditer(variable_name)  function is used to directly iterate the array
    print(i)
print()
for i,d in np.ndenumerate(var):          # ndenumerate(variable_name) is show index and data together 
    print(i,d)
```
